# **2. Estudio de Predictores de Diabetes en la Población Estadounidense 2022**
## **2.1. Definición del Problema**

La diabetes es una de las enfermedades crónicas más prevalentes en Estados Unidos y representa un riesgo significativo para la salud pública. Su detección temprana permite intervenir antes de que se desarrollen complicaciones graves. Sin embargo, la gran diversidad demográfica, conductual y clínica de la población dificulta construir modelos predictivos robustos, especialmente cuando los casos positivos son minoritarios.
El reto principal es desarrollar un sistema predictivo que identifique de forma confiable a los individuos con mayor riesgo de presentar diabetes utilizando datos de salud poblacional.

## **2.2. Objetivo General**

Desarrollar un modelo de Machine Learning capaz de predecir la presencia de diabetes en adultos de Estados Unidos durante 2022, maximizando la sensibilidad hacia la clase positiva y utilizando un proceso metodológico completo que abarque desde la caracterización inicial de los datos hasta la validación del modelo final.

## **2.3. Objetivos Específicos**

- Analizar la estructura, distribución y características de las variables incluidas en el dataset.

- Identificar factores demográficos, conductuales y clínicos asociados con la diabetes.

- Implementar técnicas de preprocesamiento adecuadas para datos mixtos (numéricos y categóricos).

- Entrenar y evaluar múltiples modelos tradicionales de clasificación para establecer líneas base.

- Abordar el desbalance significativo entre clases mediante técnicas estadísticas y algorítmicas.

- Diseñar y validar un modelo predictivo original que supere las limitaciones encontradas en los métodos estándar.

## **2.4. Justificación**

El dataset Heart 2022 contiene información detallada sobre el estado de salud y condiciones de vida de más de 246.000 adultos. Su variable objetivo, HadDiabetes, se encuentra altamente desbalanceada, lo que dificulta la construcción de modelos capaces de identificar casos positivos con precisión aceptable.
Por ello, el proyecto es pertinente por tres razones:

- Relevancia social y sanitaria: detectar diabetes de forma temprana tiene impacto directo en la prevención de enfermedades crónicas, hospitalizaciones y costos en salud.

- Relevancia técnica: los datasets desbalanceados representan uno de los retos más comunes y complejos en clasificación binaria.

- Relevancia académica: permite aplicar el ciclo completo de Machine Learning, evaluar limitaciones reales y explorar modelos más allá de los algoritmos convencionales.

**Dataset:** Personal Key Indicators of Heart Disease – 2022

**Archivo utilizado:** heart_2022_no_nans.csv

**Registros:** 246.022

**Variables:** 40

**Valores faltantes:** No contiene

**Fuente:**
https://www.kaggle.com/datasets/kamilpytlak/personal-key-indicators-of-heart-disease?select=2022

## **2.6. Descripción General del Dataset**

El dataset incluye variables de tipo:

**Numéricas (6):** hábitos de sueño, estado físico y mental, medidas corporales, IMC.

**Categóricas (34):** antecedentes médicos, actividades diarias, factores socio-demográficos, discapacidades, estilo de vida, vacunación y más.

**La variable objetivo es:**

``HadDiabetes (0 = No, 1 = Sí)``

**Clase positiva:** ~13.7%, lo que genera un problema claro de desbalance.

## **2.7. Metodología del Proyecto**

El proyecto se desarrolló siguiendo un flujo metodológico estructurado:

### **1. Análisis Exploratorio de Datos (EDA)**

- Descripción estadística de variables.
- Visualización de distribuciones.
- Identificación de correlaciones y patrones relevantes.
- Detección de valores atípicos y evaluación de calidad del dato.

### **2. Preprocesamiento**

- Codificación de variables categóricas.
- Estandarización de variables numéricas.
- Manejo de desbalance mediante distintas estrategias.
- Preparación de datasets limpio y sin atípicos.

### **3. Modelado Baseline**

Entrenamiento de modelos tradicionales para establecer un punto de comparación:

- Regresión Logística
- Random Forest
- Gradient Boosting
- Variantes con class_weight y técnicas como SMOTE/SMOTENC

### **4. Evaluación**

Se utilizaron métricas como:

- Recall (enfoque principal)
- F1-score
- Accuracy
- ROC–AUC
- Matriz de confusión

### **5. Optimización**

- Búsqueda de hiperparámetros.
- Ensayos con diferentes configuraciones para maximizar sensibilidad.

### **6. Propuesta del Modelo Original**

Se diseñó un modelo híbrido personalizado (HCSE model) para superar las limitaciones observadas en técnicas estándar, priorizando escalabilidad y sensibilidad. (Los detalles del modelo se presentan en secciones posteriores del informe.)

### **7. Validación Final**

- Evaluación del rendimiento del modelo original.
- Comparación con líneas base.
- Discusión de resultados y análisis crítico.

## **2.8. Alcance del Proyecto**

El proyecto se centra en predicción y diagnóstico preliminar, no en inferencia causal.
No pretende reemplazar evaluaciones médicas, sino servir como herramienta analítica para apoyar sistemas de salud o estudios poblacionales.

## **3. Variables**

El dataset contiene 40 variables que se clasifican en dos grandes grupos:

### **Variables numéricas (6)**
- **`PhysicalHealthDays`**: Número de días en el último mes con mala salud física.  
- **`MentalHealthDays`**: Número de días en el último mes con mala salud mental.  
- **`SleepHours`**: Horas promedio de sueño por noche.  
- **`HeightInMeters`**: Altura en metros.  
- **`WeightInKilograms`**: Peso en kilogramos.  
- **`BMI`**: Índice de Masa Corporal.

### **Variables Categóticas (34)**
- **`State`**: Estado de residencia en EE. UU.  
- **`Sex`**: Sexo del encuestado.  
- **`GeneralHealth`**: Autopercepción del estado de salud general.  
- **`LastCheckupTime`**: Tiempo desde la última revisión médica.  
- **`PhysicalActivities`**: Realiza actividad física regularmente (Sí/No).  
- **`RemovedTeeth`**: Número de dientes extraídos.  
- **`HadHeartAttack`**, **`HadAngina`**, **`HadStroke`**, **`HadAsthma`**, **`HadSkinCancer`**, **`HadCOPD`**, **`HadDepressiveDisorder`**, **`HadKidneyDisease`**, **`HadArthritis`**, **`HadDiabetes`**: Antecedentes de condiciones de salud (Sí/No).  
- **`DeafOrHardOfHearing`**, **`BlindOrVisionDifficulty`**: Discapacidades sensoriales (auditiva y visual).  
- **`DifficultyConcentrating`**, **`DifficultyWalking`**, **`DifficultyDressingBathing`**, **`DifficultyErrands`**: Dificultades funcionales en tareas diarias (Sí/No).  
- **`SmokerStatus`**: Estado de fumador (Nunca, Ex-fumador, Fumador).  
- **`ECigaretteUsage`**: Uso de cigarrillos electrónicos (Sí/No).  
- **`ChestScan`**: Realización de escáner de tórax (Sí/No).  
- **`RaceEthnicityCategory`**: Categoría de raza o etnia.  
- **`AgeCategory`**: Categoría de edad.  
- **`AlcoholDrinkers`**: Consumo de alcohol en el último mes (Sí/No).  
- **`HIVTesting`**: Prueba de VIH alguna vez (Sí/No).  
- **`FluVaxLast12`**: Vacunación contra la gripe en los últimos 12 meses (Sí/No).  
- **`PneumoVaxEver`**: Vacunación antineumocócica alguna vez (Sí/No).  
- **`TetanusLast10Tdap`**: Vacunación de tétanos (Tdap) en los últimos 10 años (Sí/No).  
- **`HighRiskLastYear`**: Exposición a entornos de alto riesgo COVID-19 en el último año (Sí/No).  
- **`CovidPos`**: Resultado positivo de COVID-19 en el último año (Sí/No).
